```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
# -------------------------------------------------
# 1. Read a low-contrast grayscale image
# -------------------------------------------------
# Replace with your own image path
img = cv2.imread(r"C:\Users\20241CAI0023\Downloads\ryan-johns-hDWrqoCDAdk-unsplash.jpg", cv2.IMREAD_GRAYSCALE)
if img is None:
    # Create a synthetic low-contrast image for testing
    img = np.random.randint(80, 140, (300, 400), dtype=np.uint8)
    cv2.imwrite('low_contrast_synthetic.jpg', img)
# -------------------------------------------------
# 2. Apply Global Histogram Equalization
# -------------------------------------------------
equ = cv2.equalizeHist(img)
# -------------------------------------------------
# 3. Compute histograms
# -------------------------------------------------
hist_original = cv2.calcHist([img], [0], None, [256], [0, 256])
hist_equalized = cv2.calcHist([equ], [0], None, [256], [0, 256])
# -------------------------------------------------
# 4. Display results
# -------------------------------------------------
plt.figure(figsize=(14, 8))
# Original Image
plt.subplot(2, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Low-Contrast Image')
plt.axis('off')
# Equalized Image
plt.subplot(2, 2, 2)
plt.imshow(equ, cmap='gray')
plt.title('After Histogram Equalization')
plt.axis('off')
# Original Histogram
plt.subplot(2, 2, 3)
plt.plot(hist_original, color='blue')
plt.title('Original Histogram')
plt.xlabel('Intensity')
plt.ylabel('Frequency')
plt.xlim([0, 256])
plt.grid(True, alpha=0.3)
# Equalized Histogram
plt.subplot(2, 2, 4)
plt.plot(hist_equalized, color='green')
plt.title('Equalized Histogram')
plt.xlabel('Intensity')
plt.ylabel('Frequency')
plt.xlim([0, 256])
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('histogram_equalization_result.png', dpi=150, bbox_inches='tight')
plt.show()
# Optional: Save the enhanced image
cv2.imwrite('equalized_output.jpg', equ)
print("Enhanced image saved as 'equalized_output.jpg'")
```
